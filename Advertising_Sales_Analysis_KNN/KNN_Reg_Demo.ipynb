```python
cat_cols = df.select_dtypes(include="object").columns
num_cols = df.select_dtypes(exclude="object").columns

print("Categorical Columns:")
print(cat_cols)

print("\nNumerical Columns:")
print(num_cols)

```

```python
X_cat = df[cat_cols]
X_num = df[num_cols]
y = df["Sales"]
```

```python
X_cat
```

```python
# Fit encoder
from sklearn.preprocessing import OneHotEncoder
import pickle
encoder = OneHotEncoder(drop="first", handle_unknown="ignore")
X_cat_encoded = encoder.fit_transform(X_cat)
pickle.dump(encoder, open("encoder.pkl", "wb"))
```

```python
X_cat
```

```python
X_num = df[num_cols]
```

```python
X_num.drop("Sales", axis=1, inplace=True)
```

```python
X_num.head()
```

```python
y.head()
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_num_scaled = scaler.fit_transform(X_num)
pickle.dump(scaler, open("scaler_knn.pkl", "wb"))
```

```python
X_final = np.hstack((X_cat_encoded.toarray(), X_num_scaled))
```

```python
X_final[0:5]
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(
    X_final,   # encoded + numeric data 
    y,
    test_size=0.2,
    random_state=42
)
```

```python
print("Training X shape:", X_train.shape)
print("Testing X shape:", X_test.shape)
print("Training y shape:", y_train.shape)
print("Testing y shape:", y_test.shape)
```

```python
from sklearn.neighbors import KNeighborsRegressor

knn = KNeighborsRegressor(
    n_neighbors=7,
    weights="distance",
    metric="minkowski",
    algorithm="auto",
    n_jobs=-1,
    p=1
)
knn.fit(X_train, y_train)
```

```python
y_pred = knn.predict(X_test)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error

print("MAE:", mean_absolute_error(y_test, y_pred))
print("RMSE:", np.sqrt(mean_squared_error(y_test, y_pred)))
```

```python
plt.figure()
plt.scatter(y_test, y_pred)

# Perfect prediction line
plt.plot([y_test.min(), y_test.max()],
         [y_test.min(), y_test.max()],
         linestyle="--")
''' 
Starts at (min actual, min actual)
Ends at (max actual, max actual)
Equation: Predicted = Actual
This line shows perfect prediction.
'''

plt.xlabel("Actual Sales")
plt.ylabel("Predicted Sales")
plt.title("Actual vs Predicted Sales")
plt.show()
```

```python
errors = []

for k in range(3, 40):
    knn = KNeighborsRegressor(n_neighbors=k)
    
    knn.fit(X_train, y_train)
    
    y_pred = knn.predict(X_test)
    
    rmse = np.sqrt(mean_squared_error(y_test, y_pred))
    errors.append(rmse)

plt.figure()
plt.plot(range(3, 40), errors)
plt.xlabel("K Value")
plt.ylabel("RMSE")
plt.title("Choosing Optimal K")
plt.show()
```

```python
with open("knn_sales_model.pkl", "wb") as f:
    pickle.dump(knn, f)
```

```python
from sklearn.model_selection import GridSearchCV
param_grid = {
    "n_neighbors": list(range(12, 50)),
    "weights": ["uniform", "distance"],
    'p' : [1, 2],
    "algorithm": ["auto", "ball_tree"]

}

```

```python
grid = GridSearchCV(
    estimator=knn,
    param_grid=param_grid,
    scoring="neg_root_mean_squared_error",
    cv=5
)
```

```python
grid.fit(X_train, y_train)
```

```python
best_knn = grid.best_estimator_

print("Best Parameters:", grid.best_params_)
print("Best RMSE:", -grid.best_score_)
#“Manual tuning checks few values.
#GridSearchCV checks all logical combinations and guarantees the best one.”
```

```python
knn = KNeighborsRegressor(
    n_neighbors=12,
    weights="distance",
    metric="minkowski",
    algorithm="auto",
    p=1
)
knn.fit(X_train, y_train)
```

```python
y_pred = best_knn.predict(X_test)

rmse = np.sqrt(mean_squared_error(y_test, y_pred))
print("Test RMSE:", rmse)
```

```python
with open("knn_sales_model.pkl", "wb") as f:
    pickle.dump(knn, f)
```
